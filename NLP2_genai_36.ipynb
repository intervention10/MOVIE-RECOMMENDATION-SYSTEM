```python
"""the code snippet defines and uses the extract_names() function to solve two main problems in movie recommendation datasets: 
parsing text-formatted data and cleaning tags for feature extraction."""


import ast    #ast module parses strings containing python literals (strings,lists,dicts)  into python object.
def extract_names(obj_str,key="name",max_items=None):  #obj_str: The raw string representation of a JSON-like/Python list of dictionaries (e.g., '[{"name": "Action"}, {"name": "Adventure"}]').
    try:                                               #max_items=None: Optional limit on how many items to process from the list (e.g., top 3 actors). If None, it processes all items.
        items=ast.literal_eval(obj_str)
        if not isinstance(items,list):                 #checks if items is of list data-type,if not it returns an empty list.
            return []
        return[
        item[key].replace(" ","").lower()             #item[key] -> item[1463]  -> name 'culture clash'
        for item in items[:max_items]                 #Iterates through the elements in items up to max_items
        if isinstance(item,dict)
        and key in item
        and isinstance(item[key],str)
        ]
    except(ValueError,SyntaxError):
        return []       #Returns an empty list if parsing fails, preventing the notebook from crashing on bad data or missing values.

```

```python
"""
key="name": Default key to extract from each dictionary. Defaults to "name". its default. No relation with {id,name} in our dictionary

try:
Starts a try block to handle potential errors if obj_str is malformed or invalid Python syntax.

items = ast.literal_eval(obj_str)
Converts the string obj_str into a real Python data structure (usually a list of dictionaries).

if not isinstance(items, list):
is a built-in type-checking function that checks whether the variable items is an instance of the list data type. Returns boolean true or false

return []
If items is not a list (e.g., if it parsed into None or a dictionary), return an empty list immediately.

return [
Starts a list comprehension to construct and return the final list of strings.

item[key].replace(" ", "").lower()
For each dictionary, it takes the value stored under key, removes all spaces (.replace(" ", "")), and converts it to lowercase (.lower()).
> Example: "Science Fiction" becomes "sciencefiction". (This is standard practice in recommendation systems so multi-word tags like "Sam Worthington" aren't split into separate words).

for item in items[:max_items]
Iterates through the elements in items up to max_items. If max_items is None, items[:None] simply slices the whole list.

if isinstance(item, dict) and key in item and isinstance(item[key], str)
A safety filter ensuring that:

item is actually a dictionary.

The target key exists inside item.

The value at item[key] is actually a string before calling .replace() and .lower().

]
Closes the list comprehension expression.

except (ValueError, SyntaxError):
Catches errors thrown if ast.literal_eval() fails to parse an invalid string.

return []
Returns an empty list if parsing fails, preventing the notebook from crashing on bad data or missing values.

Returning an empty list [] instead of a text message like "parsing failed" is a deliberate design choice that prevents your entire 
data pipeline from breaking down.

Here is why returning an empty list is the better approach in data science workflows:

Data Type Consistency
When you apply this function across a pandas DataFrame column:
Pandas expects every output of that function to be of the same data type (in this case, a list).

If it returns []: Every row in df['genres'] becomes a Python list (e.g., ['action', 'adventure'] or []).

Downstream Operations Expect Iterables
Later in a recommendation system pipeline, you will perform operations on these columns that assume every row is a list or iterable
"""    
```

```python
import ast
def extract_director(crew_str):
    try:
        crew=ast.literal_eval(crew_str) #Converts the raw string crew_str into an actual Python list of dictionaries, where each dictionary represents a crew member.

        for member in crew:
            if member.get('job') == 'Director': #Using .get('job') is safer than member['job'] because it won't crash if the 'job' key is missing from a dictionary.
                return[member['name'].replace(" ","").lower()]
    except:
        return []
    return []


```

```python
"""Why use .get('job') instead of member['job']?                  

Direct Access (member['job'])                                       Safe Access (member.get('job'))                                      
If a crew dictionary is missing the 'job' key entirely, 
Python crashes with a KeyError.                                     If the 'job' key is missing, it returns None, making None == 'Director'
                                                                    evaluate safely to False.
                                                                    
Requires extra if 'job' in member: checks to prevent crashes.       Handles missing keys cleanly in a single line."""    
```

```python
"""CountVectorizer is a natural language processing (NLP) tool used to convert text data into numerical vectors using the 
Bag-of-Words (BoW) model.

Why it's used here:
Computers cannot directly compare raw text strings (like your combined movie tags) to calculate mathematical similarity. 
CountVectorizer prepares your data for vector math by:

Building a Vocabulary: Collecting unique words across all movie tag strings in the dataset.

Counting Frequencies: Converting each movie's tag string into a numerical vector of word counts.

How it leads to the next cells:
Cell [157]: You instantiate cv = CountVectorizer(max_features=5000, stop_words='english'). This selects the top 5,000 most 
frequent words across all movies while removing standard English filler words (like "the", "is", "at").

Cell [159]: You pass your text strings to cv.fit_transform(final_df['tags']), turning every movie into a 5,000-dimensional
numerical vector.

Cell [163]: You use those numerical vectors with cosine_similarity to measure how close (similar) two movies are in 
5,000-dimensional space."""
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
"""Cosine similarity is a metric used to measure how similar two vectors (or text documents) are, regardless of their size or 
length.Instead of measuring the straight-line distance between two data points, it measures the angle theta between their 
vectors in a multi-dimensional space.

The result always falls between -1 and 1 (or 0 to 1 for positive frequency vectors like text counts):
1-> theta = 0 deg: The vectors point in the exact same direction. The texts are identical in word proportions.
0-> theta = 90 deg: The vectors are orthogonal (perpendicular). The two movies share zero words in common.
-1-> theta = 180 deg: The vectors point in opposite directions (not applicable when using count frequencies, as word counts 
cannot be negative)

3. Why Cosine Similarity over Euclidean Distance?   
In recommendation systems using Bag-of-Words:

Euclidean Distance measures the absolute distance between points. If Movie A has a 50-word tag description and Movie B has a 
500-word description covering the exact same plot, Euclidean distance will rank them as far apart simply because Movie B is longer.

Cosine Similarity focuses purely on orientation/direction. Since it only cares about the angle, a short description and a long
detailed description covering similar themes will still yield a similarity score close to 1.

"""
```

```python
cv=CountVectorizer(max_features=5000,stop_words='english')

#cv = CountVectorizer(max_features=5000, stop_words='english'). This selects the top 5,000 most 
# frequent words across all movies while removing standard English filler words (like "the", "is", "at").
 
"""CountVectorizer: Converts a collection of text documents to a matrix of token counts. It scans all texts, builds a vocabulary 
 of known words, and counts how many times each word appears in each movie's tag.

max_features=5000: Restricts the vocabulary to only the top 5,000 most frequently occurring words across the entire dataset. 
This keeps the vector size manageable, reduces memory usage, and filters out ultra-rare noise words 
(like typos or obscure character names).

stop_words='english': Automatically removes standard English "stop words" (such as "the", "is", "at", "and", "in"). Since these 
common words carry little to no semantic meaning, removing them prevents them from skewing the similarity calculations.
"""
```

```python
vectors=cv.fit_transform(final_df['tags']).toarray()

"""cv.fit_transform(final_df['tags']):

fit(): Learns the vocabulary—it scans every text entry in final_df['tags'] and identifies the top 5,000 most frequent unique
words (excluding English stop words).

transform(): Converts every movie's tag string into a numerical vector based on that learned 5,000-word vocabulary, counting 
word occurrences for each movie.

.toarray(): By default, fit_transform() returns a SciPy sparse matrix (to save memory since most entries in the matrix are 
zeros). Calling .toarray() converts that sparse matrix into a standard 2D NumPy array."""
```

```python
#---------------------------MOVIE RECOMMENDATION ENGINE-------------------NLP------------


import os
os.chdir("D:\\allstudymaterials\\GenAI")

import pickle
import ast
import numpy as np
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity

movies=pd.read_csv("tmdb_5000_movies.csv")
credits=pd.read_csv("tmdb_5000_credits.csv")

df=movies.merge(credits,on='title')
df=df[['id','genres','keywords','overview','title','cast','crew']]
# df.isnull().sum()
df['overview']=(df['overview']).ffill()
# df.isnull().sum()

 
def extract_names(obj_str,key='name',max_items=None):
    try:
        items=ast.literal_eval(obj_str)
        if not isinstance(items,list):
            return[]
        return[
        item[key].replace(" ","").lower()
        for item in items[:max_items]
        if isinstance(item,dict)
        and key in item
        and isinstance(item[key],str)
        ]
    except(ValueError,SyntaxError):
        return[]

df['genres']=df['genres'].apply(extract_names)
df['keywords']=df['keywords'].apply(extract_names)
df['cast']=df['cast'].apply(lambda x:extract_names(x,max_items=3))

def extract_director(crew_str):
    try:
        crew=ast.literal_eval(crew_str)
        
        for member in crew:
            if member.get('job')=='Director':
                return[member['name'].replace(" ","").lower()]
    except:
        return []
    return []

df['crew']=df['crew'].apply(extract_director)
df['overview']=df['overview'].astype(str).str.lower().str.split()
df['tags']=df['overview']+df['keywords']+df['genres']+df['crew']+df['cast']

# df['tags'][0]
df['tags']=df['tags'].apply(lambda x:" ".join(x))
final_df=df[['id','title','tags']]



cv=CountVectorizer(max_features=5000,stop_words='english')
vectors=cv.fit_transform(final_df['tags']).toarray()
similarity=cosine_similarity(vectors)

def recommend(movie):
    movie_index=movies[movies['title']==movie].index[0]
    distances=similarity[movie_index]

    movies_list=sorted(
    list(enumerate(distances)),
    reverse=True,key=lambda x:x[1])[1:10]
    for i in movies_list:
        print(movies.iloc[i[0]].title)

pickle.dump(final_df.to_dict(),open('movies_dict.pkl','wb'))
pickle.dump(similarity,open('similarity.pkl','wb'))
                       
    
# recommend('Avatar')
```

```python
#--------------------------------STREAMLIT APP-----------------------------------------------------


import streamlit as st
import pandas as pd
import pickle

# ---------------------------------------------------------
# Page Config & Setup
# ---------------------------------------------------------
st.set_page_config(page_title="Movie Recommender System", layout="centered")
st.title("Movie Recommender System")

# ---------------------------------------------------------
# Load Precomputed Files (Cached)
# ---------------------------------------------------------
@st.cache_data
def load_data():
    movies_dict = pickle.load(open('movies_dict.pkl', 'rb'))
    movies = pd.DataFrame(movies_dict)
    similarity = pickle.load(open('similarity.pkl', 'rb'))
    return movies, similarity

movies, similarity = load_data()

# ---------------------------------------------------------
# Recommendation Function
# ---------------------------------------------------------
def recommend(movie):
    movie_index = movies[movies['title'] == movie].index[0]
    distances = similarity[movie_index]
    movies_list = sorted(list(enumerate(distances)), reverse=True, key=lambda x: x[1])[1:10]
    
    recommended_movies = []
    for i in movies_list:
        recommended_movies.append(movies.iloc[i[0]].title)
    return recommended_movies

# ---------------------------------------------------------
# UI Layout
# ---------------------------------------------------------
selected_movie = st.selectbox(
    "Select or type a movie to get recommendations:",
    movies['title'].values
)

if st.button('Recommend'):
    recommendations = recommend(selected_movie)
    st.subheader(f"Top 9 recommendations for **{selected_movie}**:")
    
    for idx, rec_movie in enumerate(recommendations, 1):
        st.write(f"{idx}. {rec_movie}")
```
